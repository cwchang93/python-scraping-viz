## check_q10


```python
def check_q10(driver, base_url, run):
    path = Path("output/exercise_sql.png")
    path.parent.mkdir(exist_ok=True)
    path.unlink(missing_ok=True)
    returned = run(path)
    assert Path(returned) == path, "要回傳 path"
    assert path.exists() and path.stat().st_size > 1000, "截圖檔不存在或太小"
    assert driver.find_element(By.ID, "count").text == "共 1 筆", "搜尋 sql 應該有 1 筆結果"
```

## main


```python
def main():
    checks = [
        ("1", "首頁標題", lambda d, b: q1_home_title(d, b) == "練習站"),
        ("2", "JavaScript 清單數量", lambda d, b: q2_count_js_items(d, b) == 6),
        ("3", "導覽列文字", lambda d, b: q3_nav_texts(d, b) == ["首頁", "登入", "會員頁", "搜尋", "延遲載入"]),
        (
            "4",
            "搜尋結果數量",
            lambda d, b: (q4_search_count(d, b, "python"), q4_search_count(d, b, "git"), q4_search_count(d, b, "zzz"))
            == (5, 1, 0),
        ),
        ("5", "登入成功", lambda d, b: (q5_login_welcome(d, b) or "").startswith("歡迎回來")),
        ("6", "錯誤密碼的訊息", lambda d, b: q6_wrong_password_message(d, b) == "帳號或密碼錯誤"),
        ("7", "訂單編號", lambda d, b: q7_order_ids(d, b) == ["A-1001", "A-1002", "A-1003", "A-1004", "A-1005"]),
        ("8", "明確等待", lambda d, b: q8_wait_for_result(d, b) == "資料載入完成：共 3 筆"),
        ("9", "渲染後解析", lambda d, b: q9_delayed_items(d, b) == ["Alpha", "Beta", "Gamma"]),
        ("10", "搜尋並截圖", lambda d, b: check_q10(d, b, lambda path: q10_search_screenshot(d, b, path)) is None),
    ]

    server, base_url = start_server()
    driver = None
    failed = 0
    try:
        driver = new_driver()
        for number, description, check in checks:
            driver.get("about:blank")  # 每題都從空白頁開始，前一題未完成的換頁不會影響這一題
            driver.delete_all_cookies()
            started = time.time()
            try:
                passed = check(driver, base_url)
                passed = True if passed is None else bool(passed)
                note = ""
            except Exception as error:  # noqa: BLE001 - show every kind of mistake
                passed = False
                note = f"{type(error).__name__}: {str(error).splitlines()[0] if str(error) else ''}"
            failed += 0 if passed else 1
            print(f"第 {number:>2} 題 {'PASS' if passed else 'FAIL'}  {description}  ({time.time() - started:.1f}s) {note}")
    finally:
        if driver is not None:
            driver.quit()
        stop_server(server)

    print(f"\n通過 {len(checks) - failed} / {len(checks)} 題")
    sys.exit(1 if failed else 0)
```
