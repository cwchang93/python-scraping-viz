## check_q9


```python
def check_q9(d, b):
    df = q9_posts_dataframe(d, b)
    if not (isinstance(df, pd.DataFrame)):
        return False
    if not (list(df.columns) == ["id", "title", "tag", "views"]):
        return False
    if not (df.shape == (60, 4)):
        return False
    if not (df["id"].is_unique and df["id"].dtype.kind == "i"):
        return False
    if not (int(df["views"].sum()) == sum(post["views"] for post in POSTS)):
        return False
```

```python
def main():
    checks = [
        ("1", "第一批文章數", lambda d, b: q1_initial_count(d, b) == 10),
        ("2", "捲到底一次", lambda d, b: q2_scroll_once(d, b) == 20),
        ("3", "捲到全部載入", lambda d, b: q3_scroll_all(d, b) == 60),
        ("4", "最後一篇標題", lambda d, b: q4_last_title(d, b) == "第 60 篇：Matplotlib 繪圖"),
        ("5", "按到按鈕消失", lambda d, b: q5_click_load_more(d, b) == 25),
        ("6", "XPath 找進階課程", lambda d, b: q6_xpath_titles(d, b) == ["機器學習入門", "專題工作坊"]),
        ("7", "XPath 找按鈕並按一次", lambda d, b: q7_xpath_button(d, b) == 20),
        ("8", "找到目標就停", lambda d, b: check_q8(d, b) is not False),
        ("9", "整理成 DataFrame", lambda d, b: check_q9(d, b) is not False),
        ("10", "改用 API", lambda d, b: q10_titles_by_api(d, b) == EXPECTED_TITLES),
    ]

    server, base_url = start_server()
    driver = None
    failed = 0
    try:
        driver = new_driver()
        for number, description, check in checks:
            driver.get("about:blank")  # 每題都從空白頁開始，前一題未完成的動作不會影響這一題
            driver.delete_all_cookies()
            started = time.time()
            try:
                passed = check(driver, base_url)
                passed = True if passed is None else bool(passed)
                note = ""
            except Exception as error:  # noqa: BLE001 - show every kind of mistake
                passed = False
                note = f"{type(error).__name__}: {str(error).splitlines()[0] if str(error) else ''}"
            failed += 0 if passed else 1
            print(f"第 {number:>2} 題 {'PASS' if passed else 'FAIL'}  {description}  ({time.time() - started:.1f}s) {note}")
    finally:
        if driver is not None:
            driver.quit()
        stop_server(server)

    print(f"\n通過 {len(checks) - failed} / {len(checks)} 題")
    sys.exit(1 if failed else 0)
```
