## 補充題｜PTT 熱門新聞有幾則提到黃仁勳？

用一次 `requests.get()` 取得 [PTT 網頁版的今日熱門新聞](https://www.pttweb.cc/hot/news/today)，只看這次回應裡的首批清單。

1. 執行第一格，確認 HTTP 狀態碼，並觀察抓到幾個標題。這格會把 `response.text` 按標題的 HTML 標記切開，將標題整理成 `news_titles` 清單。
2. 在第二格用 `for` 逐一讀取標題。用 `if keyword in title` 判斷，將符合的標題加入 `matching_titles`，再用 `len()` 計數並列出標題。

每則貼文算一筆，包含 `Re:` 回文。清單會更新，答案以當次抓到的標題為準。換關鍵字時，只要重跑第二格。

若收到 403 或 429，停止這次練習。

```python
from html import unescape

import requests

news_url = "https://www.pttweb.cc/hot/news/today"
news_titles = []

try:
    news_response = requests.get(news_url, timeout=10)
    print("HTTP 狀態碼：", news_response.status_code)
    news_response.raise_for_status()
except requests.RequestException as error:
    print("這次未取得新聞頁面：", error)
else:
    html = news_response.text

    # PTT 網頁版的桌面標題由兩個 span 裡的文字組成。
    title_marker = '<span class="e7-show-if-device-is-not-xs"'
    suffix_marker = '<span class="__e7-full-title-appended-part"'
    for block in html.split(title_marker)[1:]:
        title_parts = block.split("</span>", 2)
        if len(title_parts) < 3 or not title_parts[1].startswith(suffix_marker):
            continue

        first_part = title_parts[0].split(">", 2)[-1]
        second_part = title_parts[1].split(">", 1)[-1]
        title = unescape(first_part + second_part).strip()
        if title:
            news_titles.append(title)

    if news_titles:
        print("首批文章數：", len(news_titles))
    else:
        print("這次未從 HTML 找到新聞標題，請先檢查回應內容。")
```
